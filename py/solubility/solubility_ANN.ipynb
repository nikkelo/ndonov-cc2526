```python
import pandas as pd
import matplotlib.pyplot as plt
import rdkit, rdkit.Chem
import numpy as np
import tensorflow as tf
```

import data

```python
df = pd.read_csv("https://github.com/whitead/dmol-book/raw/main/data/curated-solubility-dataset.csv")
df
```

define features

```python
X_start = list(df.columns).index("MolWt")
X_names = df.columns[X_start:]
X = df[X_names]
y = df['Solubility']
```

linear regression

```python
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.33, random_state = 42)

scaler = preprocessing.StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

lr = LinearRegression()
lr.fit(X = X_train, y = y_train)

y_pred = lr.predict(X_test)
print("R2 score: " + str(r2_score(y_test,y_pred)))
```

ANN

```python
model = tf.keras.models.Sequential([
tf.keras.Input(shape=(17,)),
tf.keras.layers.Dense(20, activation='relu'),
tf.keras.layers.Dense(20, activation='relu'),
tf.keras.layers.Dense(1)
])

tf.keras.utils.plot_model(model,
show_shapes=True,
show_dtype=True,
show_layer_names=True,
rankdir='TB',
expand_nested=True,
dpi=64,
show_layer_activations=True,
show_trainable=True
)

model.compile(optimizer='adam', loss='mse')

model.fit(np.array(X_train), np.array(y_train), batch_size = 16, epochs = 50)

r2_metric = tf.keras.metrics.R2Score()

r2_metric.update_state(np.array(y_test), model.predict(np.array(X_test)))

print("R2 Score:", r2_metric.result().numpy())
```

do the same thing, but use rdkit to featurize the molecules instead

```python
mols = pd.read_csv("https://github.com/whitead/dmol-book/raw/main/data/curated-solubility-dataset.csv")
molecules = [rdkit.Chem.MolFromSmiles(smi) for smi in mols.SMILES]

valid_mol_idx = [bool(m) for m in molecules]
valid_mols = [m for m in molecules if m]
len(valid_mols)
```

```python
from rdkit.Chem import Descriptors

vals = [Descriptors.CalcMolDescriptors(x) for x in valid_mols]
X = pd.DataFrame(vals)
X
```

use descriptors to train model instead

```python
X.dropna(inplace = True, axis = 1)
len(X), len(X.columns)

X = np.clip(X.values.astype(np.float64), -1000, 1000)
X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.33, random_state = 42)

scaler = preprocessing.StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

model = tf.keras.models.Sequential([
tf.keras.Input(shape=(205,)),
tf.keras.layers.Dense(50, activation='relu'),
tf.keras.layers.Dense(50, activation='relu'),
tf.keras.layers.Dense(50, activation='relu'),
tf.keras.layers.Dense(1)
])

tf.keras.utils.plot_model(model,
show_shapes=True,
show_dtype=True,
show_layer_names=True,
rankdir='TB',
expand_nested=True,
dpi=64,
show_layer_activations=True,
show_trainable=True
)

model.compile(optimizer='adam', loss='mse')

model.fit(np.array(X_train), np.array(y_train), batch_size = 16, epochs = 50)

r2_metric = tf.keras.metrics.R2Score()

r2_metric.update_state(np.array(y_test), model.predict(np.array(X_test)))

print("R2 Score:", r2_metric.result().numpy())
```
